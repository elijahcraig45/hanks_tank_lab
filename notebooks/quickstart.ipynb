## BigQuery with the guards
`%%bq NAME` runs the SQL (dry-run first, capped scan, capped result, cached) and stores the DataFrame as NAME.

```python
%%bq games --max-gb 1
SELECT status, COUNT(*) AS n
FROM `hankstank.mlb_2026_season.games`
GROUP BY status ORDER BY n DESC
```

## The registry
Versions are immutable and hash-verified; channels are pointers.

```python
from mllab.registry import Registry
reg = Registry('/mnt/appdata/mllab/registry')
print('versions:', reg.versions('hankstank', 'game_outcome'))
print('channels:', reg.channels('hankstank', 'game_outcome'))
```

## Your experiments
Each run gets its own verdict; runs are never ranked against each other because they use different games.

```python
from mllab.jobs import LocalRunner
from mllab import runs
print(runs.render_runs(runs.list_runs(LocalRunner('/mnt/appdata/mllab'), limit=5)))
```

## Model card

```python
print(runs.card(reg, LocalRunner('/mnt/appdata/mllab'), 'hankstank', 'game_outcome', 'v10'))
```
